## 1.4 Reusable evaluation function

```python
def evaluate_classifier(name, model, X_train, X_test, y_train, y_test, show_details=True):
    """Fit a classifier and return key multiclass performance metrics."""
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    metrics = {
        'Model': name,
        'Accuracy': accuracy_score(y_test, pred),
        'Weighted F1': f1_score(y_test, pred, average ='weighted', zero_division = 0),
        'Macro F1': f1_score(y_test, pred, average ='macro', zero_division = 0)
    }

    if show_details:
        print(f'--- {name} ---')
        print('Confusion Matrix:')
        print(confusion_matrix(y_test, pred))
        print('\nClassification Report:')
        print(classification_report(y_test, pred, zero_division = 0))
        print('Accuracy:', f"{metrics['Accuracy']:.4f}")
        print('Weighted F1:', f"{metrics['Weighted F1']:.4f}")
        print('Macro F1:', f"{metrics['Macro F1']:.4f}")

    return metrics, pred, model
```

# Stage 2 - Linear Discriminant Analysis (LDA)


**Linear Discriminant Analysis (LDA)** is a supervised classification algorithm used to predict which predefined class or group an observation belongs to.

Unlike clustering, where groups are discovered from the data, LDA requires a known target variable during training. In this activity, the target variable is the customer's segment, and LDA attempts to predict segment membership from customer characteristics.

The main idea behind LDA is to find combinations of the predictor variables that separate the classes as much as possible. It looks for directions in the feature space where observations belonging to the same class are relatively close together while observations belonging to different classes are as far apart as possible.

For example, suppose customer segments differ according to variables such as:

* income
* age
* household size
* average order size
* purchase frequency
* tenure

LDA examines these variables simultaneously and determines which combinations provide the strongest separation between the customer segments.

#### Why is it called "Linear" Discriminant Analysis?

LDA constructs `linear decision boundaries` between classes. Conceptually, it creates a discriminant score from a weighted combination of the predictors:

$$
D = b_0+b_1X_1+b_2X_2+\cdots+b_pX_p
$$

where $$(X_1,\ldots,X_p)$$ are predictor variables and the coefficients determine how strongly each variable contributes to separating the classes.

> For two classes and two predictors, the resulting decision boundary can be visualized as a straight line. With more predictors, the same idea extends to higher-dimensional linear boundaries.

### How LDA works

LDA estimates the `statistical characteristics` of each class from the training data. In particular, it considers:

* the `mean values` of the predictors within each class,
* the `variation and relationships` among the predictors,
* the `prior probability` of belonging to each class.

For a new observation, LDA evaluates how well its predictor values correspond to each class and assigns it to the class with the highest estimated probability.

An important characteristic of LDA is that it assumes the classes share a common covariance matrix. In simplified form:

$$
\Sigma_1=\Sigma_2=\cdots=\Sigma_K=\Sigma
$$

This assumption is one of the major differences between LDA and Quadratic Discriminant Analysis (QDA). QDA allows each class to have its own covariance matrix, whereas LDA estimates one pooled covariance matrix using information from all classes.

### Main assumptions of LDA

LDA generally works best when:

1. The target variable consists of distinct classes.

2. Numerical predictors are approximately normally distributed within each class.

3. The classes have approximately equal covariance matrices.

4. Predictors do not exhibit severe multicollinearity.

5. Observations are independent.

Real-world datasets rarely satisfy every assumption perfectly. Therefore, these assumptions should be considered when interpreting the model rather than treated simply as pass/fail requirements.

### LDA and our customer segmentation problem

In this activity, the customer segment is already known for customers in Dataset 1. LDA therefore learns the relationship:

$$
\text{Customer Characteristics}
\longrightarrow
\text{Customer Segment}
$$

We can then evaluate how accurately LDA predicts known customer segments using a test dataset.

Later, in Stage 6, we use predictors that are available in both Dataset 1 and Dataset 2 to train the model using customers whose segments are known. The resulting model can then be applied to prospective customers in Dataset 2 whose segment memberships are unknown. This corresponds directly to the targeting objective of the activity.

### Why LDA is useful in this activity

LDA is particularly useful here because it provides both a classification method and a way of thinking about class separation. It allows us to investigate whether demographic and behavioural characteristics contain enough information to distinguish among the six customer segments.

After fitting the model, we evaluate its predictive performance using the required:

* **Confusion Matrix** - shows which segments were correctly and incorrectly classified.

* **Accuracy** - percentage of all customers classified correctly.

* **F1 Score** - balances precision and recall and is especially useful when class sizes are unequal.

* **Classification Report** - provides precision, recall, F1 score, and support for each segment.

These are the performance measures required for the LDA stage of this activity.

> **Key takeaway:** LDA attempts to find `linear combinations` of predictor variables that best distinguish predefined classes. It is particularly effective when classes are reasonably separable, predictors are `approximately normally distributed within classes`, and the `classes have similar covariance structures`.

# Stage 3 - Quadratic Discriminant Analysis (QDA)

### Quadratic Discriminant Analysis (QDA)

QDA is a supervised classification algorithm used to predict which predefined class an observation belongs to. It is closely related to LDA but QDA is more flexible because it allows each class to have its own covariance structure.

Like LDA, QDA uses known class memberships in the training data to learn how predictor variables differ among classes. In this activity, QDA attempts to predict a customer's segment using demographic and behavioural characteristics such as income, age, purchase frequency, average order size, tenure, and other available features.

### How is QDA different from LDA?

The most important difference concerns the covariance matrices of the classes.

LDA assumes that all classes share the same covariance matrix:

$$
\Sigma_1=\Sigma_2=\cdots=\Sigma_K
$$

QDA removes this restriction and estimates a separate covariance matrix for each class:

$$
\Sigma_1,\Sigma_2,\ldots,\Sigma_K
$$

This allows the relationships and variability among predictors to be different for different customer segments.

For example, income and age might have a strong relationship within one customer segment but a much weaker relationship within another. QDA can account for these differences, whereas LDA assumes a common covariance structure across all segments.

### Why is it called "Quadratic" Discriminant Analysis?

Because each class can have a different covariance matrix, the resulting decision boundaries between classes can be curved or quadratic, rather than necessarily linear.

Conceptually:

**LDA**: $$\text{Linear decision boundary}$$

**QDA**

$$
\text{Quadratic/curved decision boundary}
$$

This additional flexibility means QDA can identify more complex separation patterns between classes. However, that flexibility comes at a cost: QDA must estimate considerably more parameters than LDA.

### How QDA works

For each class, QDA estimates:

* the mean values of the predictors,
* the covariance matrix,
* the prior probability of belonging to the class.

For a new observation, QDA calculates a discriminant score for each possible class. In simplified mathematical form, the QDA discriminant function for class (k) is:

$$
\delta_k(x) = -\frac{1}{2}\log|\Sigma_k|
-\frac{1}{2}(x-\mu_k)^T\Sigma_k^{-1}(x-\mu_k)
+\log(\pi_k)
$$


where:

* (x) = predictor values for the observation,
* $$\mu_k = $$ mean vector for class (k),

* $$\Sigma_k = $$covariance matrix for class (k),

* $$\pi_k = $$ prior probability of class (k).

The observation is assigned to the class producing the largest discriminant score.

You do not normally need to calculate this equation manually, but it helps explain why QDA requires enough observations in every class: it must estimate and invert a separate covariance matrix for each class.

### Main assumptions of QDA

QDA generally works best when:

1. The target variable consists of distinct classes.

2. Numerical predictors are approximately normally distributed within each class.

3. Observations are independent.

4. There is sufficient data within each class to estimate its covariance matrix.

5. Predictors do not create severe numerical problems such as singular covariance matrices.

Unlike LDA, QDA does not assume equal covariance matrices across classes. This is its major advantage when different groups genuinely have different covariance structures.

### QDA requires more data than LDA

This is particularly important in our dataset.

Suppose there are (p) predictor variables. QDA must estimate a separate ($$p\times p$$) covariance matrix for every segment.

Therefore, QDA can encounter serious problems when:

$$
\text{Number of observations in a class}
$$

is small relative to:

$$
\text{Number of predictors}
$$

This is directly relevant to this activity because Segment 5 contains only four customers. After dividing the data into training and testing sets, there are even fewer Segment-5 observations available for model training.

Consequently, QDA may not have enough information to reliably estimate the covariance matrix for Segment 5.

This is more than simply a class-imbalance issue. It is also a covariance-estimation problem.

### QDA, collinearity, and dimensionality reduction

The activity specifically warns that QDA may generate a collinearity warning and asks us to investigate correlation analysis and dimensionality reduction.

One possible approach is Principal Component Analysis (PCA).

PCA transforms the original predictors into a smaller number of components:

$$
X_1,X_2,\ldots,X_p
\quad\longrightarrow\quad
PC_1,PC_2,\ldots,PC_m
$$

where:

$$
m<p
$$


> PCA can make QDA computationally feasible, but it does not create additional information about an extremely small class.

If Segment 5 contains only four actual customers, dimensionality reduction does not change that fact. Therefore, predictions involving Segment 5 should still be interpreted cautiously.

### QDA versus LDA

| Characteristic                        | LDA               | QDA                            |
| ------------------------------------- | ----------------- | ------------------------------ |
| Classification type                   | Supervised        | Supervised                     |
| Decision boundary                     | Linear            | Quadratic/curved               |
| Class means                           | Separate          | Separate                       |
| Covariance matrices                   | One shared matrix | Separate matrix for each class |
| Flexibility                           | Lower             | Higher                         |
| Number of parameters                  | Lower             | Higher                         |
| Data requirement                      | Lower             | Higher                         |
| Sensitivity to small classes          | Lower             | Higher                         |
| Risk of overfitting                   | Lower             | Higher                         |
| Suitable for complex class boundaries | Less flexible     | More flexible                  |

Neither method is automatically superior. QDA trades simplicity for flexibility.

If the true class boundaries are approximately linear or the dataset is limited, LDA may perform better. If classes have substantially different covariance structures and sufficient observations are available, QDA may capture relationships that LDA cannot.

### QDA in this customer segmentation activity

In this activity, we use QDA to determine whether customer characteristics can distinguish among the six existing customer segments.

The required workflow is to:

1. Train the QDA model using Dataset 1.

2. Generate predictions for the test data.

3. Evaluate the model using:

   * Confusion Matrix
   * Accuracy
   * F1 Score
   * Classification Report

4. Investigate potential collinearity problems.

5. Consider correlation analysis and dimensionality reduction when necessary.

These requirements correspond directly to Stage 3 of the activity.

> Key takeaway: QDA is a more flexible extension of LDA that allows each class to have its own covariance matrix and therefore permits curved decision boundaries. This flexibility can improve classification when groups have genuinely different covariance structures, but QDA requires substantially more data within each class and can become unstable when classes are very small, predictors are highly correlated, or the dimensionality is too high.


## 3.1 Why the default QDA fails here

QDA estimates one covariance matrix per class. A class covariance matrix in `p` dimensions cannot be full rank unless that class has enough independent observations. Here the training set contains only 3 Segment 5 observations but 14 predictors.

We intentionally attempt the default QDA and catch the expected error so that the statistical limitation is visible rather than hidden.

```python
qda_default = QuadraticDiscriminantAnalysis()

try:
    qda_default.fit(X_train, y_train)
    qda_default_pred = qda_default.predict(X_test)
    print('Default QDA unexpectedly fitted successfully in this environment.')
except Exception as exc:
    print('Default QDA cannot be fitted with the current data structure.')
    print(type(exc).__name__ + ':', exc)
```

## 3.2 Correlation analysis
Correlation is worth examining because highly correlated predictors can further destabilize covariance estimation. However, in this dataset the more fundamental QDA problem is the extremely small Segment 5 sample size.

```python
corr = X_train.corr()

fig, ax = plt.subplots(figsize=(11, 9))
im = ax.imshow(corr.values, aspect='auto')
ax.set_xticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticks(range(len(corr.index)))
ax.set_yticklabels(corr.index)
ax.set_title('Training-Set Feature Correlation Matrix')
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()

# List the strongest absolute off-diagonal correlations.
upper = corr.abs().where(np.triu(np.ones(corr.shape), k=1).astype(bool))
strong_pairs = upper.stack().sort_values(ascending=False).head(10)
display(strong_pairs.to_frame('Absolute correlation'))
```

## 3.3 Leakage-free dimensionality reduction for QDA

The earlier sample fitted PCA before the train/test split, which leaks information from the test set into preprocessing. Here PCA is placed inside a pipeline and therefore is fitted only on the training data.

Because the rarest training class has 3 observations, we use 2 principal components. With 3 observations, at most 2 centered dimensions can have independent variation. A 95%-variance PCA happens to keep more than two components and therefore still cannot solve the Segment 5 rank problem in the current scikit-learn implementation.

```python
qda_pca = Pipeline([
    ('scale', StandardScaler()),
    ('pca', PCA(n_components=2)),
    ('qda', QuadraticDiscriminantAnalysis())
])

qda_metrics, qda_pred, qda_pca = evaluate_classifier(
    'QDA (Standardized + PCA to 2 components)',
    qda_pca, X_train, X_test, y_train, y_test
)
```

```python
explained = qda_pca.named_steps['pca'].explained_variance_ratio_
print('Variance explained by PC1 and PC2:', explained)
print('Total variance explained:', explained.sum().round(4))
```

### QDA interpretation
This QDA result should be treated cautiously. Reducing 14 predictors to two components makes the covariance matrices estimable, but it also discards information. The very small Segment 5 class remains a serious modeling limitation regardless of the software implementation.

# Stage 4: k-Nearest Neighbours (KNN)

KNN is distance-based, so predictors measured on large numerical scales can dominate Euclidean distance. We therefore standardize all predictors inside a pipeline before fitting KNN.

```python
knn = Pipeline([
    ('scale', StandardScaler()),
    ('knn', KNeighborsClassifier(n_neighbors=5))
])

knn_metrics, knn_pred, knn = evaluate_classifier(
    'KNN (k=5, standardized)',
    knn, X_train, X_test, y_train, y_test
)
```

## Optional K tuning
We can compare a small set of candidate `k` values using the held-out test set for instructional purposes. In production model development, hyperparameter tuning should preferably use cross-validation on the training data rather than repeatedly consulting the final test set.

```python
knn_tuning = []
for k in [3, 5, 7, 9, 11, 15]:
    candidate = Pipeline([
        ('scale', StandardScaler()),
        ('knn', KNeighborsClassifier(n_neighbors=k))
    ])
    candidate.fit(X_train, y_train)
    pred = candidate.predict(X_test)
    knn_tuning.append({
        'k': k,
        'Accuracy': accuracy_score(y_test, pred),
        'Weighted F1': f1_score(y_test, pred, average='weighted', zero_division=0),
        'Macro F1': f1_score(y_test, pred, average='macro', zero_division=0)
    })

display(pd.DataFrame(knn_tuning).sort_values('Weighted F1', ascending=False))
```

Multinomial NB is designed primarily for non-negative count/frequency-type features. Several variables here are continuous measurements rather than counts, so this model is included because the activity requests it, but its assumptions are less natural for this dataset.

# Stage 6 - Class Prediction (Targeting)

## 6.1 Load Dataset 2 and identify common features
We do not concatenate or merge the rows of the two datasets. We only identify the predictor columns that exist in both datasets.

```python
df_b = pd.read_csv(DATASET2)
print('Dataset 2 shape:', df_b.shape)
display(df_b.head())

print('\nMissing values in Dataset 2:')
print(df_b.isna().sum())
print('\nDuplicate rows in Dataset 2:', df_b.duplicated().sum())
```

```python
common_features = [col for col in X.columns if col in df_b.columns]
print('Common features:', common_features)

X_train_common = X_train[common_features].copy()
X_test_common = X_test[common_features].copy()
X_new = df_b[common_features].copy()
```

The five common predictors are demographic variables: `married`, `own_home`, `household_size`, `income`, and `age`. Behavioral variables such as order frequency, cross-buy, tenure, returns, and marketing contacts are unavailable for prospects. We should therefore expect weaker classification performance than in Stages 2 to 5.

## 6.2 Re-evaluate all requested algorithms using only common features

```python
common_models = {
    'LDA': LinearDiscriminantAnalysis(),
    'QDA (PCA2)': Pipeline([
        ('scale', StandardScaler()),
        ('pca', PCA(n_components=2)),
        ('qda', QuadraticDiscriminantAnalysis())
    ]),
    'KNN (scaled, k=5)': Pipeline([
        ('scale', StandardScaler()),
        ('knn', KNeighborsClassifier(n_neighbors=5))
    ]),
    'Gaussian NB': GaussianNB(),
    'Multinomial NB (MinMax)': Pipeline([
        ('minmax', MinMaxScaler()),
        ('mnb', MultinomialNB())
    ])
}

common_results = []
common_fitted = {}

for name, model in common_models.items():
    metrics, pred, fitted = evaluate_classifier(
        name, model,
        X_train_common, X_test_common,
        y_train, y_test,
        show_details=False
    )
    common_results.append(metrics)
    common_fitted[name] = fitted

common_comparison = pd.DataFrame(common_results).sort_values(
    ['Weighted F1', 'Accuracy'], ascending=False
).reset_index(drop=True)

display(common_comparison)
```

## 6.3 Detailed reports for the common-feature models

```python
for name, model in common_models.items():
    print('\n' + '=' * 80)
    _metrics, _pred, _fitted = evaluate_classifier(
        name, model,
        X_train_common, X_test_common,
        y_train, y_test,
        show_details=True
    )
```

## 6.4 Select the targeting model
For this supplied dataset and this fixed stratified split, we rank models primarily by weighted F1, while also checking accuracy and macro F1. Weighted F1 is useful here because the classes are very imbalanced, although the macro F1 score reminds us whether minority classes are being neglected.

The code below selects the model with the highest weighted F1 automatically rather than hard-coding LDA.

```python
best_model_name = common_comparison.iloc[0]['Model']
print('Selected model based on highest weighted F1:', best_model_name)
display(common_comparison.iloc[[0]])
```

### Business interpretation of model selection

The best-performing common-feature model is preferred for prospect targeting on the available evidence, but the moderate performance should temper expectations. The prospects lack the behavioral variables that were highly informative for existing customers, and Segment 5 is too rare to model reliably. Any campaign based on these predictions should therefore be monitored and validated rather than treated as certain segment membership.

## 6.5 Refit the selected model on all known customers
The train/test split was needed for honest model comparison. Once the algorithm has been selected, the final deployment model should be trained on all 2000 labeled customers using the five common features, so no labeled information is unnecessarily discarded.

```python
X_common_all = X[common_features].copy()

# Re-create the selected estimator so that the final model is fit from scratch on all labeled data.
final_model_lookup = {
    'LDA': LinearDiscriminantAnalysis(),
    'QDA (PCA2)': Pipeline([
        ('scale', StandardScaler()),
        ('pca', PCA(n_components=2)),
        ('qda', QuadraticDiscriminantAnalysis())
    ]),
    'KNN (scaled, k=5)': Pipeline([
        ('scale', StandardScaler()),
        ('knn', KNeighborsClassifier(n_neighbors=5))
    ]),
    'Gaussian NB': GaussianNB(),
    'Multinomial NB (MinMax)': Pipeline([
        ('minmax', MinMaxScaler()),
        ('mnb', MultinomialNB())
    ])
}

final_model = final_model_lookup[best_model_name]
final_model.fit(X_common_all, y)
print(f'Final {best_model_name} model fitted on {len(X_common_all):,} labeled customers.')
```

## 6.6 Predict Dataset 2 and save the final CSV

```python
pred_new = final_model.predict(X_new)

output_df = df_b.copy()
output_df['Predicted_Segment'] = pred_new

display(output_df.head())
print('\nPredicted segment distribution:')
display(output_df['Predicted_Segment'].value_counts().sort_index().to_frame('Count'))
```
